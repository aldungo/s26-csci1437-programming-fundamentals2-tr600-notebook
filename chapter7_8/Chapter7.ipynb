### Another Powerful Example: Find Biggest/Smallest Number in Array

Step Approach:
1. Assume the 0th position has the smallest/largest number 
2. Go through every number in the array and check if it is bigger or smaller than the "current value" 
    * if so, overwrite and proceed checking every element in the arrayu

```java
int[] numbers = {5, 9, 6, 3, 7, 0, 4, 2, 8, 1};

int largestNumber = numbers[0]; // assume the 0th position is the LARGEST number

for (int index = 1; index < numbers.length; index++) // start comparing against 1ST position, since 0th is already saved
{
    // check which number is greater
    if (numbers[index]  > largestNumber) // check if current element we're on is BIGGER than our largestNumber
        largestNumber = numbers[index]; // 9 > 5
        // if so, reassign the value of our largestNumber to b the current element value
}

System.out.println("The largest number is " + largestNumber);
```

The above example can be expanded on a tiny bit further by also keeping track of the LOCATION (index) of the biggest element. 


```java
int[] numbers = {5, 9, 6, 3, 7, 0, 4, 2, 8, 1};

int smallestNumber = numbers[0]; // assume the 0th position is the SMALLEST number
int smallestIndex = 0;
for (int index = 1; index < numbers.length; index++) // start comparing against 1ST position, since 0th is already saved
{
    // check which number is smaller
    if (numbers[index]  < smallestNumber) // check if current element we're on is SMALLER than our smallestNumber
    {
        smallestNumber = numbers[index]; // 
        // if so, reassign the value of our smallestNumber to b the current element value
        smallestIndex = index; // while reassigning the value of smallestNumber, also keep track of where it's at
    }
}
System.out.println("The smallest number is " + smallestNumber + " at location " + smallestIndex);
```

### More Advanced Array Processing

One useful action to be performed on array elements is to SWAP values.

In the below example, we'll swap the first and last values of the array. 

```java
int[] numbers = {5, 9, 6, 3, 7, 0, 4, 2, 8, 1};

System.out.println("BEFORE");
for (int index = 0; index < numbers.length; index++) 
{
    System.out.print(numbers[index] + " ");
}

// first attempt at swapping...
numbers[0] = numbers[9]; // put the value 1 (which is at position 9) at the 0th position
// numbers[0] is now 1
numbers[9] = numbers[0]; // put the value 5 (which is at position 0) at the 9th position

/** THIS IS NOT THE RIGHT WAY TO SWAP!! */

System.out.println("\nAFTER");
for (int index = 0; index < numbers.length; index++) 
{
    System.out.print(numbers[index] + " ");
}
```

The above doesn't work because statements process one at a time. So overwriting didn't work and the second assignment does not perform anything meaningful. 

To FIX this, a third variable TEMPORARILY holds the value that gets overwritten. 

```java
int[] numbers = {5, 9, 6, 3, 7, 0, 4, 2, 8, 1};

System.out.println("BEFORE");
for (int index = 0; index < numbers.length; index++) 
{
    System.out.print(numbers[index] + " ");
}

int temp = numbers[0];
numbers[0] = numbers[9]; // put the value 1 (which is at position 9) at the 0th position
numbers[9] = temp; 

System.out.println("\nAFTER");
for (int index = 0; index < numbers.length; index++) 
{
    System.out.print(numbers[index] + " ");
}
```

### Another Example: Randomly Shuffle Array Elements

```java
int[] numbers = {5, 9, 6, 3, 7, 0, 4, 2, 8, 1};

System.out.println("BEFORE");
for (int index = 0; index < numbers.length; index++) 
{
    System.out.print(numbers[index] + " ");
}

for (int index = 0; index < numbers.length; index++)
{
    // for loop allows me to go through each array element
    // for each iteration...

    int randomIndex = (int)(Math.random() * numbers.length); // generate a random number between 0 - [arraySize] - 1
    int temp = numbers[index]; // use temp variable to save value that's about to get overwritten
    numbers[index] = numbers[randomIndex];
    numbers[randomIndex] = temp;
}

System.out.println("\nAFTER");
for (int index = 0; index < numbers.length; index++) 
{
    System.out.print(numbers[index] + " ");
}
```

### Shift Array Elements to the Left

```java
int[] numbers = {5, 9, 6, 3, 7, 0, 4, 2, 8, 1}; // {9, 6, 3, 7, 0, 4, 2, 8, 1, 5};

System.out.println("BEFORE");
for (int index = 0; index < numbers.length; index++) 
{
    System.out.print(numbers[index] + " ");
}

int temp = numbers[0]; // save the value at the beginning (0th index)
for (int index = 1; index < numbers.length; index++)
{
    numbers[index - 1] = numbers[index];
}

numbers[numbers.length - 1] = temp;

System.out.println("\nAFTER");
for (int index = 0; index < numbers.length; index++) 
{
    System.out.print(numbers[index] + " ");
}
```

### Copying Arrays

Arrays are objects (reference variables, pointing to an objec tlocation, not the value) so the reference variables are what holds arrays. **When you assign one variable to another one, you only copy the *reference***. 

```java
int[] numbers = {3, 1, 4, 2, 5};

int[] newNumbers = numbers; // pointing to same reference... this is dangerous, try not to do this!

numbers[2] = 10000; // both arrays affected!

for (int number : numbers)
{
    System.out.print(number + " ");
}
System.out.println();
for (int number : newNumbers)
{
    System.out.print(number + " ");
}
```

When one array's data is modified, the other array variable will see those changes too! There are **two specific** possible ways to copy arrays.

1. Use a loop to copy each individual element
2. System.arraycopy

```java
int[] numbers = {3, 1, 4, 2, 5};

/*Copy arrays the right way! */ 
int[] newNumbers = new int[numbers.length]; // actually initialize a new array object

for (int index = 0; index < newNumbers.length; index++)
{
    newNumbers[index] = numbers[index];
}


numbers[2] = 10000; 

for (int number : numbers)
{
    System.out.print(number + " ");
}

System.out.println();

for (int number : newNumbers)
{
    System.out.print(number + " ");
}
```

#### System.arraycopy
Takes FIVE arguments:
- array being copied
- starting position of elements to copy
- array copying to
- starting position to place elements being copied
- amount of elements being copied

```java
int[] numbers = {3, 1, 4, 2, 5};

/*Copy arrays the right way! */ 
int[] newNumbers = new int[numbers.length]; // actually initialize a new array object

System.arraycopy(numbers, 0, newNumbers, 0, 3);

for (int number : newNumbers)
{
    System.out.print(number + " ");
}
```

### Arrays Can Be Passed to/through Methods

References (pointing to location, rather) are what's passed instead, rather than the actual values of the array. Get familiar with this, because it's the case for any argument parameter of the reference type

```java
public static void printArray(int[] arr) 
{
    for (int number : arr) {
        System.out.println(number);
    }
}
```
