# Chapter 14: JavaFX Basics

JavaFX Applications *inherit* from Application class, which has `launch` and `start` methods 
- `launch` actually launches the JavaFX Application
- `start` ABSTRACT method, that's overriden in order to create your Java application

```python
/* DON'T RUN, just for visual understanding */

import javafx.application.Application;
import javafx.stage.Stage;

public class MyFirstGUI extends Application 
{
    
    public static void main(String[] args) 
    {
        launch(args);  // Launch the JavaFX application
    }
    
    @Override
    public void start(Stage primaryStage) 
    {
        primaryStage.setTitle("My First GUI Application");
        primaryStage.show();
    }
}
```

In this chapter 14... you'll hear "theater" references/analogies 

For example, a Stage object called `primaryStage` is passed TO the start method

`Stage` class - represents the application window
`Scene` class - represents collection of **nodes** (actors) for the application 

**Nodes** - various different VISUAL elements to be placed 'on screen' in the application

The second example, `Example02_LabelScene.java` is the first introduction to adding a NODE to a scene, that is set on the application stage (window). We use a LABEL to display some text. 

### Different Types of Nodes

#### Panes
**pane** - container CLASS, that can position other nodes in a scene
* there are different types of panes, but the few that we focus on here include:
1. FlowPane - arranges nodes in a row, wrapping to next when needed
2. GridPane - arranges nodes in a GRID, with rows and columns
3. HBox - arranges nodes horizontally IN A SINGLE ROW 
4. VBox - arranges nodes vertically in a single column 

Nodes can be easily removed and added to panes, by using `ObservableList` of nodes `getChildren()` method


Nodes may include panes, UI controls, shapes, text, i.e. something you can VISUALLY see 


As we dive into more JavaFX examples, this next one `Example03_CircleShape.java` not only demonstrates using/seeing our nodes, but also nicely shows a good structure on how JavaFX Applications can be set up. 
- implement individual nodes FIRST 
- implement containers to put the nodes in (so you can place them) 
- then impelment the scene (put nodes/containers on to scene)
- appropriate syntax to then automatically launch scene onto stage 

### Properties/Methods for Nodes

- JavaFX has style properties for setting CSS
