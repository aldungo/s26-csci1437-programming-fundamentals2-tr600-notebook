## Loops
A **loop** reduces the need for duplicate lines of code, as it is a section block of code where statements execute repeatedly *for as long as a condition is true*. 

` while (boolean expression) { // lines that repeatedly execute }`

Using the repetitive example from above, let's have a print statement execute for a total of five times with a loop:

```java
int count = 1; // note how this is initialized before the loop

while (count <= 5) // while "count" is less than or equal to 5...
{
    // ... do this code repeatedly!!
    System.out.println(count + " Hello, Java!");

    count++; // THIS IS IMPORTANT! count needs to keep increasing so that it is EVENTUALLY, greater than 5 therefore making the expression FALSE 

    // when we get to the "end" of the loop, go back to the top (reiterate) until the condition finally becomes FALSE 
}

```

The **while** loop is an example of a **counter-controlled loop** ; the value of a variable count  determines how many times the loop executes

## Do-While Loop
* loop body executes AT LEAST ONCE, before the expression is checked
* in contrast to the standalone while loop, this means that even if initially false, the loop will still RUN. 



```java
do 
{
    // code to repeatedly execute
} while (boolean expression);
```

This is particularly useful if the lines of code *should execute AT LEAST ONCE*. Useful for input validation! More definition on this later, but we actually call this a **posttest loop** - which is when the Boolean expression is checked AFTER EXECUTING LOOP BODY. 

This is in contrast to what's called a **pretest loop** - which checks the Boolean expression BEFORE the loop body. 

## The For Loop
* a more concise syntax than while loops (counter-controlled loops), reducing chances for MISTAKES
*go back to 10/13's video to see off-by-one errors and infinite loops*

There are 3 parts to a for loop:

`for (initialization; boolean expression; update)`

^^^ Think, "where do i start? where do i end? how do i get there?"

initialization - where does the loop start? where are we initializing?  what is the initial value to count? 
boolean expression - determines when the loop stops 
update - what happens on each iteration of the loop? does the value being kept track of (count) update, change, go up or down? 



All of the above, separated and indicated by a semicolon `;`. Think of each of these pieces as a single statement 

```java
for (int count = 0; count < 10; count++) // 10 < 10 IS FALSE
{
    // lines to repeatedly execute
    System.out.println(count + " Hello, world!");
}
```

Now that we know all 3 loops (while, do-while, and FOR), before this chapter ends, "which loop is best to use?" That's subjective, because all three loops can technically achieve the same outcome. It's dependent on design! 

* For loops are useful for count-controlled loops 

* While loops are good for anything else, especially when the condition should be checked FIRST (pretest)

* Do-while loop can be used when statements must be executed *at least once*.

What are nested loops? Recall nested if-statements...
* Loops can be nested inside of other loops!
**Inner loop is executed COMPLETELY (all iterations) before the outer loop moves onto the next iteration.**

Total Number of Executions = # of executions per loop multiplied

```java
// 2 iterations
for (int i = 0; i < 2; i++)
{
    // code repeatedly executes
    for (int j = 0; j < 5; j++)
    {
        // inner loop executes FIVE times, for EACH iteration of the outer loop 
        System.out.println("i = " + i + ", j = " + j);
    }

}
```

There are two keywords that can be used when executing loops, but it's not really recommended. In fact, `break` and `continue` in loop is considered bad practice... Use break in switch statements!

`continue` - when used in a loop, it will terminate current iteration and move on to the next

```java
int number = 0;

while (number <= 11) // while number is less than or equal to 11... if number is 11, that's fine, iterate again! 
{
    number++;
    if (number == 9)
    {
        continue;
    }
    System.out.println("The number is " + number);
}
```

`break` - when used in a loop, it terminates the loop COMPLETE and exits.

```java
int number = 0;

while (number <= 11) // while number is less than or equal to 11... if number is 11, that's fine, iterate again! 
{
    number++;
    if (number == 9)
    {
        break;
    }
    System.out.println("The number is " + number);
}

System.out.println("The loop has ended.");
```

Remember, break and continue is bad practice! BOOLEAN EXPRESSIONS should be what CLEARLY communicates how the loop ends.
