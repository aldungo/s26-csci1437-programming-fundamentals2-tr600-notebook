## OBJECT-ORIENTED PROGRAMMING

You should have been introduced to this concept in PFI (Chapter 9 Classes and Objects)

**Class** - a .java file, more specifically a blueprint or template of what an *object* should be. A class defines what an object's characteristics and abilities are

**Object** - an instance of a class, representing a unique/distinct entity in the real world. Objects of the same type fall under the same class definition. Objects belong TO a class
- Objects have state and behaviors
- What does the object "have"? What can the object "do"? 

Analogy Example - Person Class
You can have many "person" objects... what does a person "have"?
- Name
- Birthday 

OOP is HUGE in advanced programming, because it allows for large-scale software systems to be developed.
