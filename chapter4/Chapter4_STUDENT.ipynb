One of the most commonly used Math service methods not just in Java, but programming in general is the concept of 
**RANDOMNESS**.

`Math.random()` - this alone generates a random number between `0 (inclusive) <= number < 1 (exclusive)`

```java
System.out.println(Math.random());
```

However, when COMBINED with other math expressions, we can use it to generate a number in a **specific range**.

`(int) (Math.random() * n) + m;`

* The (int) is a casting being done to an actual integer
* Multiplying allows for this *random* number to fall between 0 and (n - 1)

```java
int random = (int) (Math.random() * 11); //(int) (Math.random() * n);
// random will be any integer number between 0 and 10 

System.out.println("The random number is " + random);

int random = (int) (Math.random() * 101);// 
// random will be any integer number between 0 and 100 

System.out.println("The random number is " + random);
```

The addition of m allows for the random range to be *shifted by m*. 

```java
int random = (int) (Math.random() * 11) + 10; // this gives a random number between 0 and 10 

// shift the range by 5, so now it is a random number between 10 and 20

System.out.println("The random number is " + random);
```

All of this is to say...that because of Unicode, char values can actually be converted to numeric value equivalents. Numbers cast to characters only use LOWER 16 BITS. This allows for the conversion below to be IMPLICIT

## The String type

```java
String message = "Welcome to Java!";

int number = 23; // this variable is an integer and it is called number. Stored in memory, identified by this variable IS the value 23
```

Strings are special compared to other data types, which we discussed before (int, double, char, boolean) - these are called **primitive data types**. 

Strings are actually considered a **reference type**, a class. When you declare a string variable, you are actually referring to a String object. A String object represents a sequence of characters. 

### Instance vs. Static Methods

Up until now, for Math class and Character class, we've seen the usage of **static methods**, which are methods performed ON THE CLASS.

`Math.pow(2, 3);` - `pow` method from the Math class
`Character.isDigit(ch);` - this isDigit method comes from Character class 

Strings work differently! We still use methods to "perform and do things", but for Strings we do those methods on the actual specific objects. This is done with **instance methods**. 

* `length()` - returns the length of a String

```java
String message = "Welcome";

System.out.println("The length is " + message.length()); // length() gives us the length of the String
System.out.println("The first character is " + message.charAt(0)); // charAt() returns character at specified location 
System.out.println("The last character is " + message.charAt(message.length() - 1));
```

In programming, when we count, we start at 0! This is important to remember later on for "index" or "position". 

* trim() - takes away white space

```java
System.out.println(message.concat(" to Java!")); // concat(str) adds str to the end of a String

String uppercase = "HOW ARE YOU TODAY";
System.out.println(uppercase.toLowerCase());

uppercase = "how are you today";
System.out.println(uppercase.toUpperCase());

System.out.println("\tGood night\n");
System.out.print("hello");

System.out.println("\tGood night\n".trim());
System.out.print("hello");
```

### Comparing Strings and Checking If Equal

* Remember that Strings are REFERENCE VARIABLES.

The equality operator `==` compares references, not contents. This means that the equality operator will not work properly when comparing and checking Strings! See below:

```java
String name = "JOHN DOE";
String person = "John Doe";

// This is not the right way!!!
if (name == person.toUpperCase()) // what does person now contain? => JOHN DOE = JOHN DOE
{
    System.out.println("These variables reference the same String.");
}
else
{
    System.out.println("These variables are DIFFERENT STRINGS!");
}
```

String class has special methods that specifically do correct comparison of two Strings:
* `equals(str)` - checks if calling String and str are equal
* `equalsIgnoreCase(str)` - the same as above, just ignores any capitalization

```java
String name = "JOHN DOE";
String person = "John Doe";

System.out.println(name.equals(person.toUpperCase()));

// name variable is the calling String
// person variable is the str passed in
```

* `compareTo(str)` - compares Unicode values of calling String and str
* `compareToIgnoreCase(str)` - same as above, just ignores any capitalization

```java
String fruit1 = "apple";
String fruit2 = "banana";

// apple banana
// banana apple 

if(fruit1.compareTo(fruit2) < 0) // 
{
    System.out.println(fruit1 + " comes before " + fruit2 + " alphabetically");
}
else 
{
    System.out.println(fruit2 + " comes before " + fruit1 + " alphabetically");
}
```

### Formatting Console Output

There is a `System.out.printf()` that allows for console output to be formatted.

```java
double amount = 12618.98;
double interestRate = 0.0013;
double interest = amount * interestRate;
System.out.printf("Interest is $%.2f", interest);
```

This is using **format specifiers**, `%` is the conversion character which specificies exactly how a value should be formatted in console output. 

%[flags][width][.precision]conversion

Example `%,10.2f`
* , is the flag
    * flags provide additional formatting
    * The `,` adds a thousands separator 
    * 0 flag adds leading zeroes
    * - flag specifies left justification 
* width is 10
    * This determines minimum width, also seen as spacing
* precision is 2 (followed after the decimal)
    * determines number of digits after decimal point
    * precision is only applied to floating point numbers
    * default precision is 6
* f stands for conversion of floating-point number
    Types of Conversions:
    * `%f` for floating point numbers
    * `%d` for integer numbers
    * `%s` for Strings
    * `%n` for new line characters
    * `%%` a literal percent sign
