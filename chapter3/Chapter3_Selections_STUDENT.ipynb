### Divisibility

Divisibility of numbers can powerfully be checked using the MOD operator
Mod is the remainder of division. 
So if your mod result is 0, that means NO REMAINDER = perfectly divisible 

```java
double number = 9;

if (number % 2 == 0) // this is checking if a number is evenly divisible by 2
{
    System.out.println(number + " is even ");
}
else
{
    System.out.println(number + " is odd ");
}
```

## Nested if Statements

* if statements can be placed WITHIN other if statements (nested)
* useful if you want to check the trueness of one condition first, and then follow by another

Let's modify the above divisibility, to FIRST check that a number is not 0, before checking even or odd. 

```java
double number = 7;

if (number != 0)
{
    //nested 
    // this divisibility by 2 will not even execute if the number is 0
    System.out.println("The number is not 0");
    
    if (number % 2 == 0)
    {
        System.out.println("It is an even number.");
    }
}
```

## pH example

What is pH of a solution? This is a scale in chemistry where you determine if a solution is acidic, neutral or basic (alkaline).

* Less than 7 means ACIDIC
* 7 exactly is NEUTRAL
* Greater than 7 is BASIC 

Based on the VALUE of pH, can we print out certain statements that describe this?

```java
double pH = 7.1; // initialize this value

if(pH < 7)
{
    System.out.println("The solution is acidic.");
}
else if (pH == 7)
{
    System.out.println("The solution is neutral.");
}
else
{
    System.out.println("The solution is basic.");
}
```

## Switch Statements


**Switches** are ALTERNATIVES to if-else, if-else statements.
Simpler, easier to understand! Yes... but with some limitations.

Switches have FOUR main components. 
1. test expression is the value that is BEING COMPARED (often a variable)
    * a test expression must be a variable or a literal
    * the type must be char, byte, short, int, or String
    * CANNOT BE FLOATING-POINT DUE TO INACCURACY 
2. cases are what the value from test expression, is being compared AGAINST (always a constant)
    * cases cannot be decimals. They must be a constant or a literal! 
3. break statement - indicates the END of a case. make sure you use these accordingly!
    * 
4. default section - similar to else, only executes if all cases above were false

```java
int number = 4;

switch(number) // test expression
{
    // cases
    case 1:
        // if the case MATCHES (is true), then the code HERE executes 
        System.out.println("The number is 1.");
        break;
    case 2:
        System.out.println("The number is 2.");
        break;
    case 3:
        System.out.println("The number is 3.");
        break;
    default:
        System.out.println("You didn't choose 1-3.");
}

System.out.println("The switch has ended. See how this code still ran?");
```

## Conditional Operator

**Conditional operator** `?` CONCISELY handle if-else statements 

Your usage of it in this class may depend, but it's important to know it's the only ternary operator in Java.
* Useful for print statements and assignments 

```java
double number = 9;

if (number % 2 == 0) // this is checking if a number is evenly divisible by 2
{
    System.out.println(number + " is even ");
}
else
{
    System.out.println(number + " is odd ");
}
```

The equivalent of this code, but with a conditional operator...

```java
double num = 51;
System.out.println((num % 2 == 0) ? "num is even" : "num is odd");
```

```java
int age = 21;
String type; 

type = (age < 21) ? "minor" : "adult"; // depending on result of operation, type variable is assigned a String value of "minor" (true) or "adult" (false)

System.out.println(type);
```

## Operator Precedence and Associativity

**Operator precedence** determines the order in which *different* operations are evaluated. 

**Operator Associativity** determines the  order of operations when operations have the SAME precedence. (left to right)

```java
int result = 7 * 3 - 1; // multipllyi first, then subtract
System.out.println(result);
```

```java
// What about all the same operators?
int addition = 1 + 2 + 3; 

addition = 72 + 1 + 7 + 10;
```

Above, it is all addition. So "which happens first"? 1 + 2 FIRST, which is 3

Then add 3 + 3

That is why assignment operators occur 'LAST'
