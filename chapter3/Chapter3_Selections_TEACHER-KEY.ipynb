## Logical Operators
Logical operators combine boolean expressions: `&&` (and), `||` (or), `!` (not), `^` (xor).

```java
boolean a = true;
boolean b = false;
System.out.println("a && b: " + (a && b));
System.out.println("a || b: " + (a || b));
System.out.println("!a: " + (!a));
```

## Switch Statements
Switch statements are an alternative to multiple `if-else` statements when checking a single variable against many values.

```java
int rating = 4;
switch (rating) {
    case 1: System.out.println("Poor"); break;
    case 2: System.out.println("Fair"); break;
    case 3: System.out.println("Good"); break;
    case 4: System.out.println("Very Good"); break;
    case 5: System.out.println("Excellent"); break;
    default: System.out.println("Invalid rating");
}
```

## Conditional Operator
The conditional operator `? :` is a shorthand for `if-else` statements.

```java
int score = 85;
String result = (score >= 60) ? "Pass" : "Fail";
System.out.println(result);
```

## Operator Precedence and Associativity
Operator precedence determines the order in which operations are evaluated. Associativity determines the direction of evaluation.

```java
int result = 5 + 10 * 2; // result is 25
int result2 = 1 + 2 + 3; // result2 is 6
```
