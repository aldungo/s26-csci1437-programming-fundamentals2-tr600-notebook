## Void vs. Value-Returning Methods

Void means that nothing is returned, no return statement required. 
* With a void method, "some code executes, then terminates" 
* Tip: it is *possible* to use return in a void method, in which its purpose would be to end terminate a method early 

**return statement** - allows for output from a method 
1) terminates the method (indicates the end)
2) returns something, a value
3) REQUIRED if the method type is anything but void

**value-returning methods** - REQUIRES a return statement
- "execute some code, RETURN A VALUE, then terminate" 


TRICK FOR telling the difference between the two!
Notice how method calls are used for a void method, versus a value returning method
* Value-returning methods are treated like expressions -> they evaluate to a value
    * Notice how these can be assigned to a variable 
* Whereas, void methods are treated like statements (they simply just run and that's it)
    * these CANNOT be assigned to a variable 

Make sure we are closely understanding how these methods work in a program. We also take the time to understand how changing the way we program to using methods like these is BETTER, more useful, cleaner. But why? How?

When you call on a method, it is pushed onto the **call stack** - holds all methods that are being called on, and will execute in that order. 

Imagine a stack of plates! Methods are pushed ONTOP OF the stack, with the main method always sitting on the bottom. 

Why do this? Why "modularize code"?
* easier to maintain
* easier to debug
* more reusable!!

Take the GCD code for example... what if we were to centralize/singularize/MODULARIZE just the part that performs the calculation for GCD? And put it in its OWN METHOD? 
* code is isolated to the calculation only, so the logic is clear, clarified, and readable for ONLY THE GCD CALCULATION
* any errors would then be limited to this gcd method ONLY, improving debugging by quickly identifying source of error 
* if i wanted to use the gcd code in another probram, its easier to reuse than rewrite

### Method Overloading (Hint: Lab 11!)
Definition: when multiple methods have the same name 
* this is allowed because the PARAMETER LISTINGS change, such as amount (how many there are) or data types 
* method call is matched with its paired method definition 

Caution: it is *not* considered overloading if the method name is different 

Overloading methods allows for FLEXIBLE method calls. You can reuse a method name, but modify the DEFINITION for different arguments.
