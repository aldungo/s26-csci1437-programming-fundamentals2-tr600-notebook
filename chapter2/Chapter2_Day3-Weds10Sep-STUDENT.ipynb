# Increment/Decrement Operators

`++` increments the variable by 1
`--` decrements the variable by 1

These operators can be either prefix or postfix notation...

```java
int increasedNumber = 29;

//increasedNumber = increasedNumber + 1;
increasedNumber--;
increasedNumber--;
increasedNumber--;
increasedNumber--;

System.out.println(increasedNumber);
```

Prefix notation (before) `++increasedNumber`
is different from postfix notation (after) `increasedNumber++`

With prefix, the increment/decrement is done FIRST, then other operations...
With postfix, other operations occur first, then the increment/decrement 

```java
int a = 1;
int b = ++a + 10; // a starts as 1, 

System.out.println("a = " + a);
System.out.println("b = " + b);

System.out.println("now for postfix...");


int c = 1;
int d = c++ + 10;

System.out.println("c = " + c);
System.out.println("d = " + d);
```

# Numeric Type Conversions

* Numbers, specifically their data types, can be converted during math operations
* LARGER DATA TYPES (bigger) take PRECEDENCE over the smaller ones...

```java
byte firstByte = 10;
byte second = 20;

int sum = firstByte + second;  // int is a bigger data type, so this is okay
System.out.print(sum);
```

Above is what we call a **WIDENING CONVERSION**
* it is automatic, such as a byte converting to a int,
* short converting to long, etc. 

```java
double quotient = 5.0 / 2; 
```

Below shows **integer overflow** - see how if we try to go past the range of int, we kind of "break it" because it diverts back to the lower end of the range (negative)

```java
int intNum = 2147483647 + 1;

System.out.println(intNum);
```

To fix this, use the correct fitting data type and force it to be larger. Ensure the variable data type fits! 

```java
long intNum = 2147483647L + 1;

System.out.println(intNum);
```

On the other hand... what about the other way around? Can we go from a big number to a small number? 
This would be a **narrowing conversion**, which is something to be careful with!

```java
double number = 4; // widening
System.out.println(number);

int smallNumber = (int) 987.4; // narrowing, make the big sized data (variable) convert into a smaller size...
System.out.println(smallNumber);
```

We must be careful, because through narrowing - we can lose data!!! Narrowing is not done automatically... it must be intended with what we call a **cast operation**. 

We "cast" a double variable into a integer (explicitly)
* data can be lost, so the explicit is impoprtant
* as you can see, decimal values get truncated (lost, go away)

```java
int decimalNumber = (int) 546579.213214695879;
System.out.print(decimalNumber);
```

In what context do we often use TWO decimal places? MONEY

```java
double cost = 11.853; 

// If we wanted to ROUND this, to more accurately represent money... what would that be? 11.85

// walk through some elongated, exaggerated steps of rounding... remembering that we're emphasizing CAST OPERATOR

// if i multiply cost by 100, what do i get?
cost *= 100; 
System.out.println(cost); // cost = 1185.3

// The 3 (that we no longer want, because we are rounding to two decimal places) is now BEHIND the decimal place.

// use the cast operator to "get rid of the decimal" or remove "extra decimal places" 
cost = (int) cost;
System.out.println(cost); // cost = 1185.0

// if i want to "move" the decimal back to its original place, UNDO the multplication
cost /= 100; 
System.out.println(cost);

// we can actually COMBINE all of the above, into a SINGLE step
cost = (int) (cost * 100) / 100.0;

// This is incomplete!!! See below...
```

What if it was 11.859? What should it round to?

```java
double costRounded = 11.859;

costRounded = (int) (costRounded * 100 + 0.5) / 100.0; // notice the + 0.5 for buffer

System.out.println(costRounded);
```
