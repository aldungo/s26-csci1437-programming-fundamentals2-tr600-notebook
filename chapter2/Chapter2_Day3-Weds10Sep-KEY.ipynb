## Identifiers
Identifiers are names for variables, methods, and classes. They must follow specific rules:
- Can include letters, digits, underscores, and dollar signs
- Cannot start with a digit
- Cannot be a Java keyword
- Are case-sensitive

**Try identifying valid and invalid identifiers below.**

```java
// Valid identifiers
int myInput = 5;
double Length = 10.5;

// Invalid identifiers (uncomment to test)
// int 1Item = 3;
// int Class = 4; // 'Class' is a keyword
```

## Variables
Variables store values and must be declared before use. Their scope is limited to where they are declared.

**Try declaring and initializing variables below.**

```java
int count = 1;
double radius = 5.5;
double area = radius * radius * 3.14159;
System.out.println("Area: " + area);
```

## Assignment Expressions
Assignment uses `=` to assign values. You can use variables in expressions.

**Try modifying variable values below.**

```java
int x = 1;
x = x + 1;
System.out.println("x: " + x);

int y = (x = 2) + 3;
System.out.println("y: " + y);
```

## Constants
Constants use the `final` keyword and are typically written in uppercase.

**Try using constants in calculations.**

```java
final double PI = 3.14159;
double radius = 4.0;
double area = radius * radius * PI;
System.out.println("Area using constant PI: " + area);
```

## Numeric Operations and Literals
Java supports arithmetic operations and numeric literals.

**Try using operators and literals below.**

```java
int remainder = 7 % 3;
System.out.println("Remainder: " + remainder);

double power = Math.pow(2, 3);
System.out.println("2^3: " + power);

double quotient = 5 / 2;
System.out.println("5 / 2 (integer division): " + quotient);

double correctQuotient = 5.0 / 2;
System.out.println("5.0 / 2 (floating-point division): " + correctQuotient);
```

## Fahrenheit to Celsius Conversion
Convert Fahrenheit to Celsius using the formula:
`Celsius = (5 / 9) * (Fahrenheit – 32)`

**Try writing a conversion program below.**

```java
import java.util.Scanner;

Scanner input = new Scanner(System.in);
System.out.print("Enter Fahrenheit temperature: ");
double fahrenheit = input.nextDouble();

double celsius = (5.0 / 9) * (fahrenheit - 32);
System.out.println("Temperature in Celsius: " + celsius);
```
