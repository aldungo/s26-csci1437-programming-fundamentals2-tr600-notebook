## **Class Relationships**

We are now understanding that in a Java program, we can have multiple class files that represent said "real-world entities". And just like real-world entities, classes may be "coupled" with one another or "interact". 

Relationships EMPHASIZE how one class is "used" in another class's definition. 

* Association
* Aggregation
* Composition
* Inheritance (later, in Chapter 11)

One class defined first, then another is defined *with one field being an object of THE OTHER CLASS*. 

### Association
**Association** - binary relationship describing activity between TWO classes (classes use each other)
"classes are associated with one another"
* Association can be seen with directed actions, specified in a UML Class Diagram
* Note: this often shows "multiplicity", which specifies the amount in said associated relationship

Note: Refer to Course, Student, + Faculty example 
* "Student TAKES course"
* "Faculty TEACHES course"
* How to read multipliers?
- course has EXACTLY one faculty
- course may have 5 - 60 students
- faculty may teach 0 to 3 courses
- student may take * (at least 1) or more courses 

Both of the following relationships have what we call a **has-a relationship**. 

### Aggregation
**Aggregation** - relationship where there is an OWNERSHIP between two objects

### Composition 
**Composition** - relationship where there is a DEPENDENCE of one object on another to exist 

Examples: 
* A student HAS A name
* A student HAS AN address 




### SECURITY ISSUES with Class Relationships
We have to be careful with these class relationships as TWO major issues need to be watched.

* When we assign *references* (remember, classes are reference types!) to fields, leads to direct access outside of the class. SCARY!!
* Same with returning references - we lead to direct access outside of the class!

This can cause data leaks, errors, problems, and remember when using references, a huge issue is possibly not creating the object "correctly", you could affect the data at that location without realizing it! 

We can address this with **shallow vs. deep copies**. 

* shallow copy - only refers to the object's REFERENCE

```java
Faculty lecturer = new Faculty("John Doe", "Lecturer");
Faculty newLecturer = lecturer; 

// remember this "wrong way" from Chapter 9 of copying objects???
```

* deep copy - are an actual copy of the object ITSELF
    - original AND new object have their OWN reference
    - perform deep copies when ASSIGNING to reference fields
    - perform deep copies when RETURNING reference fields

We will do examples and try it out yourself in Lab 2 - be sure to actually create deep copies when instructed to!!!

```java
Faculty lecturer = new Faculty("John Doe", "Lecturer");
Faculty newLecturer = new Faculty(lecturer.getName(), lecturer.getPosition()); 

// remember this "wrong way" from Chapter 9 of copying objects???
```

## Wrapper Classes
**Wrapper classes** are pre-defined class libraries in Java that correspond to the primitive data types. 
* These classes "wrap" primitive types as objects
- Byte
- Short
- Integer
- Long
- Float
- Double 
- Boolean
- Character 


Each "wrapper" will treat the primitive data type as an object instead, therefore converting *to* an object with said value. 

*Hint: naming convention tip - classes are often capitalized, variables are not" 

Great analogy: "imagine putting a number into a box. the box (wrapper) gives you 'handles" to manipulate it!"






Example of construction an Integer Object:

```java
Integer x4 = Integer.valueOf(32); // uses numeric value to construct
Integer x3 = Integer.valueOf("32"); // uses String value to construct Integer object

System.out.println(x3);
System.out.println("Does x3 = x4? " + (x3 == x4));
System.out.println("Does x3 = x4? " + x3.equals(x4));
```

`.valueOf()` method may create shared objects through what's called *interning*. This is memory optimization, because references are REUSED rather than constructing multiple objects with similar values. 
* Frequently used NUMERIC VALUES are placed in a special pool of memory instead!!! 

See the power here? Although created with both a String and numeric separately, and two different variables, it recognizes that the values are the same.


TIP, DON'T FORGET!!! 
* `.equals()` compares VALUES
* `==` compares references 

```java
Double decimal = Double.valueOf(32.9);
```

In the theme of understanding that these are Wrapper Class Types, we'll also come to find that these numeric classes also have CONSTANTS that similiar replicate the actual range that the primitive data types have. 

Example `byte` = -128 to 128
Integer and Double have "ranges" too! 

```java
System.out.println(Integer.MAX_VALUE);
System.out.println(Double.MAX_VALUE);
```

You can still convert between types!
* `intValue()`
* `doubleValue()`

```java
System.out.print(Double.valueOf(12.4).intValue()); // converts to... 
```

When comparing numeric OBJECTS, we're not going to use `>`, `<` a.k.a. relational operators because those are checking actual values (primitive data types)

Instead, just like we need to compare the actual values of Strings (reference types, remember?) we would approach the same for Numeric Objects.

`compareTo()` method 
* returns 1 if calling object is GREATER than passed-in object
* return -1 if calling object is LESS than passed in object
* returns 0 if objects are equal

```java
System.out.println(Double.valueOf(12.4).compareTo(Double.valueOf(12.3))); // returns 1
System.out.println(Double.valueOf(12.4).compareTo(Double.valueOf(12.5))); // returns -1
System.out.println(Double.valueOf(12.3).compareTo(Double.valueOf(12.3))); // returns 0
```

Parse methods can convert Strings to numeric values! 
    * Method takes a String and optional base 

```java
Integer.parseInt("11"); // originally a string, but can then be turned into the numeric equivalent
```

Java has automatic conversion for some actions when converting from primitive type to object. In certain instances, this will be done automatically!!!

Primitive Type => Object is called **BOXING**

Object => Primitive Type is called **UNBOXING**

If "automatically boxed", we call this AUTOBOXING

If "automatically unboxed", we call this AUTOUNBOXING

One aspect where this is done is **assignment operations**!!!

```java
Integer autoBoxed = 100; // this automatically turns int -> Integer
Double autoBoxedDouble = 3.14; // double -> Double

// What does autounboxing look like? 
int primitiveInt = autoBoxed; // primitiveInt is an actual value 100

```
