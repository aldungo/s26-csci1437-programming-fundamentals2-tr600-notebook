## 1. JavaFX Application Structure

### Key Concepts:
- **JavaFX applications inherit from `Application` class**
- Contains two key methods:
  - `launch()` - static method for launching the application
  - `start()` - abstract method you override to create the GUI

### Basic Template:

````python
import javafx.application.Application;
import javafx.stage.Stage;

public class MyFirstGUI extends Application {
    
    public static void main(String[] args) {
        launch(args);  // Launch the JavaFX application
    }
    
    @Override
    public void start(Stage primaryStage) {
        primaryStage.setTitle("My First GUI Application");
        primaryStage.show();
    }
}
````

### Notes:
- `primaryStage` is passed automatically to `start()` method
- `Stage` represents the application window
- Must call `show()` to display the window

## 2. The Theater Analogy 🎭

JavaFX uses a **theater metaphor** to describe GUI components:

| Theater Term | JavaFX Class | Description |
|--------------|--------------|-------------|
| **Stage** | `Stage` | The application window (theater stage) |
| **Scene** | `Scene` | The collection of visual elements (the scene/set) |
| **Actors** | **Nodes** | The individual visual components |

### Node Types:
- **Panes** - Container classes to position other nodes
- **UI Controls** - Labels, buttons, text fields, etc.
- **Shapes** - Circles, rectangles, lines, etc.

### Adding a Scene with a Label:

````python
import javafx.scene.Scene;
import javafx.scene.control.Label;

public void start(Stage primaryStage) {
    Label myLabel = new Label("Welcome to JavaFX!");
    
    // myLabel is the root node of the scene
    Scene scene = new Scene(myLabel);
    
    primaryStage.setTitle("My First GUI Application");
    primaryStage.setScene(scene);
    primaryStage.show();
}
````

## 3. Creating a Circle Example

**Typical JavaFX Setup Pattern:**
1. Create individual nodes (shapes, controls)
2. Create containers (panes) and add nodes to them
3. Create scene object with the container
4. Prepare and show the stage

````python
import javafx.scene.layout.Pane;
import javafx.scene.shape.Circle;
import javafx.scene.paint.Color;

public void start(Stage primaryStage) {
    // 1. Create the node (Circle)
    Circle circle = new Circle();
    circle.setCenterX(100);
    circle.setCenterY(100);
    circle.setRadius(50);
    circle.setStroke(Color.BLACK);
    circle.setFill(Color.WHITE);
    
    // 2. Create container and add node
    Pane pane = new Pane();
    pane.getChildren().add(circle);
    
    // 3. Create scene
    Scene scene = new Scene(pane, 200, 200);
    
    // 4. Prepare stage
    primaryStage.setScene(scene);
    primaryStage.setTitle("Show Circle");
    primaryStage.show();
}
````

## 4. Styling with CSS

## 5. Layout Panes

**Panes are containers** for organizing nodes in a scene.

### Getting Children:
- Use `getChildren()` method to get an `ObservableList` of nodes
- Can `add()` or `addAll()` nodes to the list

### 5.1 FlowPane

Arranges nodes in rows, wrapping to next row when needed.

````python
import javafx.scene.layout.FlowPane;
import javafx.geometry.Insets;

FlowPane flowPane = new FlowPane();
flowPane.setPadding(new Insets(11, 12, 13, 14));
flowPane.setHgap(5);  // horizontal gap between nodes
flowPane.setVgap(5);  // vertical gap between nodes

// Add nodes
flowPane.getChildren().add(node1);
flowPane.getChildren().addAll(node2, node3);
````

### 5.2 GridPane

Arranges nodes in a grid with rows and columns.

````python
import javafx.scene.layout.GridPane;
import javafx.geometry.Pos;

GridPane gridPane = new GridPane();
gridPane.setAlignment(Pos.CENTER);
gridPane.setPadding(new Insets(11.5, 12.5, 13.5, 14.5));
gridPane.setHgap(5.5);
gridPane.setVgap(5.5);

// Add nodes at specific positions (column, row)
gridPane.add(node1, 0, 0);  // column 0, row 0
gridPane.add(node2, 1, 0);  // column 1, row 0
gridPane.add(node3, 0, 1);  // column 0, row 1
````

### 5.3 HBox (Horizontal Box)

Arranges nodes horizontally in a single row.

````python
import javafx.scene.layout.HBox;

HBox hbox = new HBox(node1, node2, node3);
// node1 on left, node2 in middle, node3 on right
````

### 5.4 VBox (Vertical Box)

Arranges nodes vertically in a single column.

````python
import javafx.scene.layout.VBox;

VBox vbox = new VBox(node1, node2, node3);
// node1 at top, node2 in middle, node3 at bottom
````

## 10. Property Binding

**Property binding** connects a target object to a source object so changes in the source automatically update the target.

### Key Concepts:
- **Target** = binding object or binding property
- **Source** = bindable/observable object
- **Observable source** is an instance of `ObservableValue` interface
- Binding properties are instances of `Property` interface

### Example: Center Circle in Pane

````python
// Bind circle center to pane dimensions
circle.centerXProperty().bind(pane.widthProperty().divide(2));
circle.centerYProperty().bind(pane.heightProperty().divide(2));
````

### Understanding Property Methods:

**Two types of getters:**
- **Value getter** - returns the actual value (e.g., `double`)
- **Property getter** - returns the property object (e.g., `DoubleProperty`)

Example:
- `getCenterX()` → returns `double` value
- `centerXProperty()` → returns `DoubleProperty` object

### Numeric Binding Properties:

Common property types:
- `DoubleProperty`
- `IntegerProperty`
- `FloatProperty`
- `LongProperty`

These support arithmetic operations that return new observable properties:
- `add()`
- `subtract()`
- `multiply()`
- `divide()`

### Unidirectional vs Bidirectional Binding

**Unidirectional:**
- Changes observed in **one direction only**
- Use `bind()` method
- Example: Circle position bound to pane size

**Bidirectional:**
- Changes observed in **both directions**
- Use `bindBidirectional()` method
- Example: Two text fields that stay synchronized

````python
// Unidirectional - circle follows pane
circle.centerXProperty().bind(pane.widthProperty().divide(2));

// Bidirectional - both update each other
textField1.textProperty().bindBidirectional(textField2.textProperty());
````

## Summary

3. Create nodes (controls, shapes, images)

6. Set scene on stage and show

### Key Classes:

- **Stage** - the window

- **Color** - color specification
- **Font** - font specification
- **Image/ImageView** - image display

### Important Concepts:
- Theater analogy (Stage → Scene → Nodes)
- CSS styling (inline or external)
- Property binding (unidirectional and bidirectional)
- ObservableList for managing children

## Teaching Notes

### Key Points to Emphasize:
1. **Theater metaphor helps students understand hierarchy**
2. **Always follow the pattern**: nodes → containers → scene → stage
3. **Property binding is powerful** but can be confusing initially
4. **CSS knowledge transfers** to JavaFX styling
5. **Start with simple examples** (Label, Circle) before complex layouts

### Common Student Mistakes:
- Forgetting to call `show()` on the stage
- Confusing value getters with property getters
- Not understanding that panes are needed for multiple nodes
- Mixing up rows/columns in GridPane
- Forgetting to set scene size or it being too small

### Live Coding Suggestions:
1. Start with minimal example (empty stage)
2. Add Label to show Scene basics
3. Add Circle with Pane to show shapes
4. Show FlowPane with multiple controls
5. Show GridPane for structured layout
6. Demonstrate property binding with resizable circle
