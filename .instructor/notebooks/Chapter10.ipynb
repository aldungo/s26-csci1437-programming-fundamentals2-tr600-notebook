## 10.2 Class Abstraction and Encapsulation

**Class abstraction** means separating:
- **WHAT** a class does (interface/public methods)
- **HOW** it does it (implementation/private details)

**Encapsulation** = hiding implementation details using `private` fields

### Example: Loan Class

````java
// Loan Class - Demonstrates abstraction and encapsulation
public class Loan {
    // Private data fields (ENCAPSULATION)
    private double annualInterestRate;
    private int numberOfYears;
    private double loanAmount;
    private java.util.Date loanDate;
    
    // No-arg constructor
    public Loan() {
        this(2.5, 1, 1000);
    }
    
    // Constructor with parameters
    public Loan(double annualInterestRate, int numberOfYears, double loanAmount) {
        this.annualInterestRate = annualInterestRate;
        this.numberOfYears = numberOfYears;
        this.loanAmount = loanAmount;
        loanDate = new java.util.Date();
    }
    
    // Getters (PUBLIC INTERFACE)
    public double getAnnualInterestRate() {
        return annualInterestRate;
    }
    
    public int getNumberOfYears() {
        return numberOfYears;
    }
    
    public double getLoanAmount() {
        return loanAmount;
    }
    
    public java.util.Date getLoanDate() {
        return loanDate;
    }
    
    // Setters (PUBLIC INTERFACE)
    public void setAnnualInterestRate(double annualInterestRate) {
        this.annualInterestRate = annualInterestRate;
    }
    
    public void setNumberOfYears(int numberOfYears) {
        this.numberOfYears = numberOfYears;
    }
    
    public void setLoanAmount(double loanAmount) {
        this.loanAmount = loanAmount;
    }
    
    // Calculate monthly payment
    public double getMonthlyPayment() {
        double monthlyInterestRate = annualInterestRate / 1200;
        double monthlyPayment = loanAmount * monthlyInterestRate / 
            (1 - (1 / Math.pow(1 + monthlyInterestRate, numberOfYears * 12)));
        return monthlyPayment;
    }
    
    // Calculate total payment
    public double getTotalPayment() {
        double totalPayment = getMonthlyPayment() * numberOfYears * 12;
        return totalPayment;
    }
}

// Test the Loan class
Loan loan1 = new Loan(2.5, 5, 1000);
System.out.println("Loan Amount: $" + loan1.getLoanAmount());
System.out.println("Interest Rate: " + loan1.getAnnualInterestRate() + "%");
System.out.println("Number of Years: " + loan1.getNumberOfYears());
System.out.println("Monthly Payment: $" + String.format("%.2f", loan1.getMonthlyPayment()));
System.out.println("Total Payment: $" + String.format("%.2f", loan1.getTotalPayment()));
````

### Key Points:
- **Private fields** = implementation details hidden
- **Public methods** = interface exposed to users
- Users don't need to know HOW monthly payment is calculated, just that they CAN get it

## 10.3 Thinking in Objects - BMI Class

**Object-Oriented Design**: Model real-world concepts as objects

### BMI (Body Mass Index) Calculator

````java
public class BMI {
    private String name;
    private int age;
    private double weight;  // in pounds
    private double height;  // in inches
    
    // Constructor
    public BMI(String name, int age, double weight, double height) {
        this.name = name;
        this.age = age;
        this.weight = weight;
        this.height = height;
    }
    
    // Constructor with metric units (kg and meters)
    public BMI(String name, double weight, double height) {
        this(name, 20, weight, height);
    }
    
    // Calculate BMI
    public double getBMI() {
        double bmi = weight * 703 / (height * height);
        return Math.round(bmi * 100) / 100.0;
    }
    
    // Get BMI status
    public String getStatus() {
        double bmi = getBMI();
        if (bmi < 18.5)
            return "Underweight";
        else if (bmi < 25)
            return "Normal";
        else if (bmi < 30)
            return "Overweight";
        else
            return "Obese";
    }
    
    public String getName() {
        return name;
    }
    
    public int getAge() {
        return age;
    }
    
    public double getWeight() {
        return weight;
    }
    
    public double getHeight() {
        return height;
    }
}

// Test BMI class
BMI person1 = new BMI("John Doe", 25, 180, 70);
System.out.println("Name: " + person1.getName());
System.out.println("Age: " + person1.getAge());
System.out.println("Weight: " + person1.getWeight() + " lbs");
System.out.println("Height: " + person1.getHeight() + " inches");
System.out.println("BMI: " + person1.getBMI());
System.out.println("Status: " + person1.getStatus());
````

### Object-Oriented Thinking:
- Model a **person's BMI** as an object with properties (name, age, weight, height)
- Encapsulate **BMI calculation** inside the class
- Provide **meaningful methods** (getStatus()) instead of just data

## 10.4 Class Relationships

Three types of relationships between classes:

### 1. Association ("uses-a")
One class uses another class temporarily

### 2. Aggregation ("has-a") - WEAK ownership
One class contains another, but both can exist independently

### 3. Composition ("has-a") - STRONG ownership  
One class contains another, owner controls lifetime

### Example: Student and Course

````java
// Course class
public class Course {
    private String courseName;
    private String[] students = new String[100];
    private int numberOfStudents;
    
    public Course(String courseName) {
        this.courseName = courseName;
    }
    
    public void addStudent(String student) {
        students[numberOfStudents] = student;
        numberOfStudents++;
    }
    
    public String[] getStudents() {
        return students;
    }
    
    public int getNumberOfStudents() {
        return numberOfStudents;
    }
    
    public String getCourseName() {
        return courseName;
    }
}

// Test Course class (AGGREGATION - students exist independently)
Course course1 = new Course("CSCI 1437 - Programming Fundamentals 2");
course1.addStudent("Alice");
course1.addStudent("Bob");
course1.addStudent("Charlie");

System.out.println("Course: " + course1.getCourseName());
System.out.println("Number of students: " + course1.getNumberOfStudents());
System.out.println("Students:");
String[] students = course1.getStudents();
for (int i = 0; i < course1.getNumberOfStudents(); i++) {
    System.out.println("  " + (i + 1) + ". " + students[i]);
}
````

### Shallow Copy vs Deep Copy

````java
// Demonstrating shallow vs deep copy
public class Person {
    private String name;
    private int age;
    
    public Person(String name, int age) {
        this.name = name;
        this.age = age;
    }
    
    public String getName() { return name; }
    public int getAge() { return age; }
    public void setAge(int age) { this.age = age; }
}

// SHALLOW COPY - just copies reference
Person person1 = new Person("Alice", 20);
Person person2 = person1;  // SHALLOW COPY!

System.out.println("Before change:");
System.out.println("person1 age: " + person1.getAge());
System.out.println("person2 age: " + person2.getAge());

person2.setAge(25);

System.out.println("\nAfter changing person2 age to 25:");
System.out.println("person1 age: " + person1.getAge());  // ALSO 25!
System.out.println("person2 age: " + person2.getAge());
System.out.println("\nBoth changed because they reference the SAME object!");

// DEEP COPY - creates new object
Person person3 = new Person(person1.getName(), person1.getAge());
person3.setAge(30);

System.out.println("\nDeep copy:");
System.out.println("person1 age: " + person1.getAge());  // Still 25
System.out.println("person3 age: " + person3.getAge());  // 30
System.out.println("Different objects, independent changes!");
````

### Key Differences:
- **Shallow Copy**: Copies reference → same object, changes affect both
- **Deep Copy**: Creates new object → independent, changes don't affect original

## 10.6 Stack Class - LIFO Data Structure

**Stack** = Last In, First Out (LIFO)
- Think of a stack of plates: add to top, remove from top
- Three main operations: **push**, **pop**, **peek**

### Stack Operations:
- **push(element)** - Add element to top
- **pop()** - Remove and return top element  
- **peek()** - Return top element WITHOUT removing
- **isEmpty()** - Check if stack is empty
- **getSize()** - Get number of elements

````java
// GenericStack class - array-based implementation
public class GenericStack {
    private Object[] elements;
    private int size = 0;
    private static final int DEFAULT_CAPACITY = 16;
    
    public GenericStack() {
        elements = new Object[DEFAULT_CAPACITY];
    }
    
    // Push element onto stack
    public void push(Object value) {
        if (size >= elements.length) {
            // Array is full, double the capacity
            Object[] temp = new Object[elements.length * 2];
            System.arraycopy(elements, 0, temp, 0, elements.length);
            elements = temp;
        }
        elements[size++] = value;
    }
    
    // Pop element from stack
    public Object pop() {
        if (isEmpty()) {
            return null;
        }
        return elements[--size];
    }
    
    // Peek at top element without removing
    public Object peek() {
        if (isEmpty()) {
            return null;
        }
        return elements[size - 1];
    }
    
    public boolean isEmpty() {
        return size == 0;
    }
    
    public int getSize() {
        return size;
    }
}

// Test Stack
GenericStack stack = new GenericStack();

System.out.println("Pushing: Java, Python, C++");
stack.push("Java");
stack.push("Python");
stack.push("C++");

System.out.println("Size: " + stack.getSize());
System.out.println("Peek: " + stack.peek());  // C++ (doesn't remove)
System.out.println("Size after peek: " + stack.getSize());

System.out.println("\nPopping all elements:");
while (!stack.isEmpty()) {
    System.out.println("  Pop: " + stack.pop());  // C++, Python, Java (LIFO!)
}

System.out.println("\nStack is now empty: " + stack.isEmpty());
````

### LIFO Visualization:
```
push("Java")     push("Python")   push("C++")      pop()           pop()
                                   
                                   [C++]  ←top                      
                 [Python]          [Python]         [Python]←top    
[Java]  ←top     [Java]            [Java]           [Java]          [Java]←top
```

**Key Point**: Last item pushed is first item popped!

## 10.7 Wrapper Classes

**Problem**: Primitives (int, double, char) are NOT objects
- Can't use them where objects are required (collections, generics)
- Don't have methods

**Solution**: Wrapper classes wrap primitives as objects

| Primitive Type | Wrapper Class |
|----------------|---------------|
| boolean        | Boolean       |
| byte           | Byte          |
| short          | Short         |
| int            | Integer       |
| long           | Long          |
| float          | Float         |
| double         | Double        |
| char           | Character     |

````java
// Creating wrapper objects

// Using constructors (older way)
Integer int1 = new Integer(5);
Double double1 = new Double(5.5);
Character char1 = new Character('A');

// Using valueOf() method (preferred)
Integer int2 = Integer.valueOf(10);
Double double2 = Double.valueOf(10.5);
Boolean bool1 = Boolean.valueOf(true);

System.out.println("Integer object: " + int2);
System.out.println("Double object: " + double2);
System.out.println("Boolean object: " + bool1);

// Getting primitive values back
int primitiveInt = int2.intValue();
double primitiveDouble = double2.doubleValue();
boolean primitiveBool = bool1.booleanValue();

System.out.println("\nBack to primitives:");
System.out.println("int: " + primitiveInt);
System.out.println("double: " + primitiveDouble);
System.out.println("boolean: " + primitiveBool);
````

### Wrapper Class Constants

````java
// Useful constants in wrapper classes

System.out.println("Integer.MAX_VALUE: " + Integer.MAX_VALUE);
System.out.println("Integer.MIN_VALUE: " + Integer.MIN_VALUE);
System.out.println("Double.MAX_VALUE: " + Double.MAX_VALUE);
System.out.println("Double.MIN_VALUE: " + Double.MIN_VALUE);
System.out.println("Character.MAX_VALUE: " + (int)Character.MAX_VALUE);
System.out.println("Character.MIN_VALUE: " + (int)Character.MIN_VALUE);
````

### Parsing Strings to Numbers

````java
// Converting strings to primitive numbers

String str1 = "123";
String str2 = "45.67";
String str3 = "true";

// Parse string to primitive
int num1 = Integer.parseInt(str1);
double num2 = Double.parseDouble(str2);
boolean bool = Boolean.parseBoolean(str3);

System.out.println("Parsed int: " + num1);
System.out.println("Parsed double: " + num2);
System.out.println("Parsed boolean: " + bool);

// Math with parsed values
System.out.println("\nMath with parsed values:");
System.out.println("num1 * 2 = " + (num1 * 2));
System.out.println("num2 / 2 = " + (num2 / 2));
````

### Converting Numbers to Strings

````java
// Converting numbers to strings

int number = 42;
double pi = 3.14159;

// Method 1: Using wrapper class toString()
String str1 = Integer.toString(number);
String str2 = Double.toString(pi);

// Method 2: Using String concatenation
String str3 = "" + number;
String str4 = "" + pi;

// Method 3: Using String.valueOf()
String str5 = String.valueOf(number);
String str6 = String.valueOf(pi);

System.out.println("Integer.toString(42): " + str1);
System.out.println("Double.toString(3.14159): " + str2);
System.out.println("\"\" + 42: " + str3);
System.out.println("String.valueOf(42): " + str5);
````

## 10.8 Automatic Boxing and Unboxing

**Autoboxing** = Automatic conversion from primitive to wrapper

**Unboxing** = Automatic conversion from wrapper to primitive

Java does this automatically since Java 5!

````java
// AUTOBOXING - primitive to wrapper (automatic)

Integer intObject = 5;  // Autoboxing! Same as: Integer.valueOf(5)
Double doubleObject = 5.5;  // Autoboxing!
Boolean boolObject = true;  // Autoboxing!

System.out.println("Autoboxing:");
System.out.println("Integer object: " + intObject);
System.out.println("Double object: " + doubleObject);
System.out.println("Boolean object: " + boolObject);

// UNBOXING - wrapper to primitive (automatic)

int primitiveInt = intObject;  // Unboxing! Same as: intObject.intValue()
double primitiveDouble = doubleObject;  // Unboxing!
boolean primitiveBool = boolObject;  // Unboxing!

System.out.println("\nUnboxing:");
System.out.println("Primitive int: " + primitiveInt);
System.out.println("Primitive double: " + primitiveDouble);
System.out.println("Primitive boolean: " + primitiveBool);
````

### Autoboxing in Action

````java
// Math operations with wrapper objects (unboxing happens automatically)

Integer num1 = 10;  // Autoboxing
Integer num2 = 20;  // Autoboxing

// Behind the scenes: unboxes to int, does math, then autoboxes back to Integer
Integer sum = num1 + num2;  
Integer product = num1 * num2;

System.out.println("num1 + num2 = " + sum);
System.out.println("num1 * num2 = " + product);

// Comparison (unboxing happens)
if (num1 < num2) {  // Unboxing for comparison
    System.out.println("num1 is less than num2");
}

// Using wrapper objects in expressions
Double price = 19.99;  // Autoboxing
Double tax = price * 0.08;  // Unboxing, calculation, then autoboxing
Double total = price + tax;  // Unboxing, calculation, then autoboxing

System.out.println("\nPrice: $" + price);
System.out.println("Tax: $" + tax);
System.out.println("Total: $" + total);
````

### When to Use Wrapper Classes?

1. **Collections** - ArrayList, HashMap, etc. require objects
```java
ArrayList<Integer> list = new ArrayList<>();  // Can't use int!
list.add(5);  // Autoboxing happens
```

2. **Generics** - Type parameters require objects
```java
public <T> void print(T value)  // T must be an object
```

3. **null values** - Primitives can't be null, wrappers can
```java
Integer value = null;  // OK
int value = null;  // ERROR!
```

## 10.10 String Class - Basics

**String** = sequence of characters

**Key property**: Strings are **IMMUTABLE** (cannot be changed)
- Any "modification" creates a NEW string

````java
// Creating strings

String str1 = "Hello";  // String literal
String str2 = new String("Hello");  // Using constructor
String str3 = "Hello";  // Same literal

// String pool - literals are stored once
System.out.println("str1 == str3: " + (str1 == str3));  // true (same object!)
System.out.println("str1 == str2: " + (str1 == str2));  // false (different objects)
System.out.println("str1.equals(str2): " + str1.equals(str2));  // true (same content)

System.out.println("\n⚠️ Always use .equals() to compare strings, not ==!");
````

### String Immutability

````java
// Demonstrating immutability

String s = "Java";
System.out.println("Original: " + s);

s = s + " Programming";  // Creates NEW string, s now references the new one
System.out.println("After concatenation: " + s);

s = s.toUpperCase();  // Creates ANOTHER new string
System.out.println("After toUpperCase: " + s);

System.out.println("\nEach operation created a NEW string object!");
System.out.println("The original \"Java\" string was never modified.");
````

### Common String Methods

````java
String text = "Hello World";

// Length and access
System.out.println("Length: " + text.length());
System.out.println("Character at index 0: " + text.charAt(0));
System.out.println("Character at index 6: " + text.charAt(6));

// Substring
System.out.println("\nSubstring(0, 5): " + text.substring(0, 5));  // "Hello"
System.out.println("Substring(6): " + text.substring(6));  // "World"

// Search
System.out.println("\nindexOf('o'): " + text.indexOf('o'));  // First occurrence
System.out.println("lastIndexOf('o'): " + text.lastIndexOf('o'));  // Last occurrence
System.out.println("indexOf(\"World\"): " + text.indexOf("World"));

// Case conversion
System.out.println("\ntoUpperCase(): " + text.toUpperCase());
System.out.println("toLowerCase(): " + text.toLowerCase());

// Trim whitespace
String padded = "  Hello  ";
System.out.println("\nBefore trim: '" + padded + "'");
System.out.println("After trim: '" + padded.trim() + "'");
````

### String Comparison

````java
String s1 = "Hello";
String s2 = "hello";
String s3 = "Hello";

// Content comparison
System.out.println("s1.equals(s3): " + s1.equals(s3));  // true
System.out.println("s1.equals(s2): " + s1.equals(s2));  // false (case-sensitive)
System.out.println("s1.equalsIgnoreCase(s2): " + s1.equalsIgnoreCase(s2));  // true

// Lexicographic comparison
System.out.println("\ns1.compareTo(s2): " + s1.compareTo(s2));  // Negative (H < h in ASCII)
System.out.println("s1.compareTo(s3): " + s1.compareTo(s3));  // 0 (equal)

// startsWith / endsWith
String filename = "report.pdf";
System.out.println("\nfilename.startsWith(\"report\"): " + filename.startsWith("report"));
System.out.println("filename.endsWith(\".pdf\"): " + filename.endsWith(".pdf"));
````

### String Replacement

````java
String message = "Hello World";

// Replace characters
String replaced1 = message.replace('o', 'a');
System.out.println("Replace 'o' with 'a': " + replaced1);

// Replace substrings
String replaced2 = message.replace("World", "Java");
System.out.println("Replace 'World' with 'Java': " + replaced2);

// Replace first occurrence
String repeated = "apple apple apple";
String replaced3 = repeated.replaceFirst("apple", "orange");
System.out.println("Replace first 'apple': " + replaced3);

System.out.println("\nRemember: Original string is UNCHANGED (immutability)");
System.out.println("Original message: " + message);
````

### Splitting Strings

````java
// Split string into array

String csv = "Alice,Bob,Charlie,Diana";
String[] names = csv.split(",");

System.out.println("CSV: " + csv);
System.out.println("Split by comma:");
for (int i = 0; i < names.length; i++) {
    System.out.println("  " + (i + 1) + ". " + names[i]);
}

// Split by whitespace
String sentence = "Java is awesome";
String[] words = sentence.split(" ");

System.out.println("\nSentence: " + sentence);
System.out.println("Split by space:");
for (String word : words) {
    System.out.println("  " + word);
}
````
