# Chapter 12 - Exception Handling


What is an **exception**? In Java programming, it is a way to SIGNAL that "something's wrong". More specifically during runtime... flagging to say that "normal expected flow is not continuing HERE." 

Remember in PF1 when we explain different errors that can "happen"?
**Runtime error** - this is when an ILLEGAL operation is performed
- when this happens, **runtime errors are thrown as exception** 

**Exception** - OBJECT that represents an error/abnormal condition

A great example you've seen before is `ArrayIndexOutOfBoundsException` 
- this occurs when the program attempts to access an array index that DOESN'T exist 
    - SNAPSHOT, PAUSE!!!! In less advanced programming, this error will cause the program to terminate 
    - Imagine you are having a very important process done on an array and suddenly THE WHOLE PROGRAM STOPS due to that one exception. Scary!!!

To have a program "terminate or act as normally and smoothly as possible" despite an error being thrown... we will be taught **Exception Handling**.

Key point with Exception Handling: When something goes "wrong" in a program, you want the handling to be done in the program that CALLED the method. Not in the method itself.

## More on the `Exception` Class Type

There is a `Throwable` class that is a PARENT of `Exception` and `Error` classes

Throwable is an INTERFACE, so we have the ability to create CUSTOM Exceptions by EXTENDING Exception or a subclass

**System Errors** - come from Error Class
- critical system errors, not easily handled by US in the program
- running out of memory, low-level problems

**Exception** - come from Exception Class
- errors caused by program or external circumstances
- EXAMPLE: issues accessing/reading/writing files 

**Runtime Exception** - exceptions that occur at runtime, coming from RunTimeException Class
* programming errors due to BAD operations
* examples might be improper input, illegal math (dividing by zero), 

**Unchecked Exceptions**
RunTimeException Class and Error Class are UNCHECKED
- these are exceptions you do NOT have to handle
RunTime Exceptions - programming errors that should be corrected IN CODE
Errors - critical system errors that can't be handled through software

**Checked Exceptions** 
All other exceptions fall under here
* They are checked, so we HAVE to handle these
* Use try-catch block and/or `throws` in method header

`throws` is used in the method header to declare that a checked exception MIGHT occur

### `finally`
`finally` keyword comes "after" try/catch block, and is code that will EXECUTE NO MATTER WHAT. Whether an exception is thrown or not
Extremely useful usage of `finally` keyword to be seen in "second" half of this Chapter.... when we handle FILES!
