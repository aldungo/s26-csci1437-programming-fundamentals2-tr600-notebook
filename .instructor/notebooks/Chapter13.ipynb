Interfaces - a type of programming construct, think of it as a "contract"
* ability or capability of doing something
* still keeping our class design in mind. 

A key benefit to interfaces, is that it allows for **multiple inheritance**. Our slide/textbook examples show that a `Chicken` class can inherit from both the Edible interface, and the Animal abstract class. 

A few rules on interfaces...
* fields must be public, static, and final
* interfaces can't directly use fields
* objects can't be constructed *from* an interface
    * remember, STATIC means it belongs to the class (interface itself) 
* data (implemmentation details) of the interface is separated 

* methods must be public or protected 
* the subclass that actually does the implementation of method, must be public 

**default methods** - the ability to provide added behaviors to an interface
- this is OPTIONAL 

### The Comparable Interface
Imagine the method capability to COMPARE something. 

Who remembers the `.compareTo()` method?
- This method comes from this `Comparable` interface
- We first see how this is used to compare STRINGS

Zoom out quite a bit, and instead of thinking about this .compareTo on STRINGS exactly...
* compareTo() has the ability to compare objects. (Strings? Integer? Rectangles? Circles?)
    * MANY classes from Java API use Comparable 
    * Strings (help sort alphabetically), Calendar, and Date 
    * This method returns a -1, 0, 1
    * EXCEPT for String method... this compares Unicode values 

```java
public interface Comparable<E>
{
    public int compareTo(E o);
}
```

```java
System.out.println(Integer.valueOf(3).compareTo(5));
System.out.println(Integer.valueOf(5).compareTo(3));
System.out.println("ABC".compareTo("XYZ"));
```

What if we override and implement a CUSTOM `compareTo` method that will work on our rectangles? 

For a class to use an interface, it is not extended but rather `implements`

Once objects have the ability to be compared, it is now also possible to SORT them. Really useful!

Interfaces are another form of **generic** programming. 

`.equals()` actually behaves very similarly to `compareTo()` 
- **BEST PRACTICE** that if you override one, you should also override the other 

### Another useful interface.... `Cloneable`
The ability to... clone objects 
* This class is actually what's called a **marker interface**, which is an interface with an EMPTY BODY
* Classes that implement Cloneable are "marked" as Cloneable 

First examples, Java API classes (existing) that use it
In reality, any subclass could implement this if they choose to 

```java
java.util.Calendar cal1 = new java.util.GregorianCalendar(2013, 2, 1);
java.util.Calendar cal3 = cal1; // just a copy?
java.util.Calendar cal2 = (java.util.Calendar)cal1.clone();

System.out.println(cal1 == cal2); //remember `==` checks REFERENCES 
System.out.println(cal1 == cal3); // same reference... scary because if cal1 changes, so will cal3
```

Chapter 13 covered abstract classes AND interfaces (a form of an abstract class). Which do I use? 

#### Interfaces
* HIGHLY ABSTRACT, means more constraints
    * BUT, lack of implementation details allows for **multiple inheritance**: means a subclass can inherit from multiple interfaces
* methods must be public abstract, default, or static
* NO constructors, no new objects
* variables must be public static final 
* best practice for interface names is an adjective or noun
    * `CloneABLE`, `ComparABLE`, `Edible` 
* better for WEAK IS-A Relationship

#### Abstract Classes
* ONLY allow for **single inheritance** - a subclass can only inherit from ONE parent
* no restrictions on methods
* no objects can be invoked, constructors are allowed (as long as they're implemented in SUBCLASS)
* no restrictions on variables
* USEFUL when there is a strong **IS-A** RELATIONSHIP

The above three, (along with everything learned so far!!!) are really important to keep in mind as you dive further into programming. We're at that point where it's no longer single files, lots of .java file classes "talking" to one another. 

Keep class-design guidelines in mind! 
* Cohesion - classes describe a SINGLE entity. Methods and fields, responsibilities etc. should logically make sense to "fulfill" a certain purpose 
* Consistency - good naming conventions, style, neatness, BE STRAIGHT FORWARD with methods and fields
* Encapsulation - data encapsulation and implementation encapsulation
    - keep data fields private (don't let things leak out of the class its intended for)
    - making things readable/modifiable WHEN NECESSARY 
    - implementation encapsulation: hide implementation details from users 
* Completeness - cover as much data/fields/methods, etc is necessary FOR the context of the problem
* Clarity - having design that MAKES sense. that can be explained and understood
    - you're really getting to that point of "software developer" mindset, so challenge yourself to consider how a class, a function, a field might be flexible and useful even outside of that direct scope you're focused on
