What is Chapter 15 about?
Chapter 15 introduces the concept of "event-handling", meaning that when "something happens", what does the code do? This might be clicking, button pressing, dragging, sliding, etc. 

Example Shown:
User clicks a button, then what happens? 

A `Button` is just an object from the `Button class`, but it is what we call an **event source object**. 
Upon the button being clicked, an event "occurs", which is technically also an object itself.
An **event** in a program is a signal that something HAPPENED 

A **handler** responds to the event... which is also an object. 

### Event-driven programming
Programming occurs reliant on "firing events". If you think about it... quite different from the way we've been running programs all this time. Compile and run, then the code executes "sequentially" as written. 
"Fire an event" is the process of creating/handling an event. 

**Handlers** (also called listeners) arean instance of an event-handling interface 

For our `OK-Cancel` JavaFX example...

First, we must implement the event handler

````python
class CancelHandlerClass implements EventHandler<ActionEvent> {
    @Override
    public void handle(ActionEvent e) {
        System.out.println("Cancel button clicked");
        System.out.println("Event source: " + e.getSource());
    }
}
````

Once implemented, almost like when we implement methods but still need to CALL on them to use... we also have to **register the event handler**

````python
// Create handler objects
OKHandlerClass handler1 = new OKHandlerClass();
CancelHandlerClass handler2 = new CancelHandlerClass();
        
// Register handlers with buttons
btOK.setOnAction(handler1);
btCancel.setOnAction(handler2);
````

Notice the `.setOnAction` method that allows the button to be HANDLED by its handler, the parameter argument passed in. 

For `Example02`, notice the key aspect that the handler classes are within the same class... That's why they begin with `private class`

**Inner (nested) classes** these are defined within other classes, which is USEFUL implementation for event handler classes.
Inner classes can erference fields and methods of the OUTER CLASS. Very useful for combining dependent classes into a primary class.,

## Anonymous Inner Classes

Sometimes we don't need to create a separate class for an event handler, especially if it's only used once. We can use **anonymous inner classes** to define and create the handler object simultaneously!

**Anonymous inner class** - a class definition WITHOUT a name that creates an object in one step

````python
/* Example of anonymous inner class - DON'T RUN */

btIncrement.setOnAction(new EventHandler<ActionEvent>() {
    @Override
    public void handle(ActionEvent e) {
        clickCount++;
        messageText.setText("Click count: " + clickCount);
    }
});
````

Notice how we:
1. Use `new EventHandler<ActionEvent>()` 
2. Define the class body with `{ ... }`
3. Override the `handle()` method
4. All in ONE statement!

This is **more concise** than creating a separate named class, but still allows access to outer class fields.

## Lambda Expressions (Modern Java!)

Lambda expressions are the MOST CONCISE way to handle events in modern Java (Java 8+). They're perfect for simple event handlers!

**Lambda expression syntax**: `(parameters) -> expression or { statements }`

For a single statement:
```java
button.setOnAction(e -> System.out.println("Clicked!"));
```

For multiple statements:
```java
button.setOnAction(e -> {
    count++;
    text.setText("Count: " + count);
});
```

````python
/* Lambda expression examples - DON'T RUN */

// Single statement - no braces needed
btExit.setOnAction(e -> System.exit(0));

// Multiple statements - use braces
btIncrement.setOnAction(e -> {
    clickCount++;
    messageText.setText("Click count: " + clickCount);
    System.out.println("Incremented to: " + clickCount);
});
````

### When to Use Each Approach?

The textbook and slides give us a decision framework:

| Approach | When to Use |
|----------|-------------|
| **Lambda expression** | Several simple functionalities (1-3 lines) |
| **Anonymous inner class** | Few slightly complex functionalities (4-10 lines) |
| **Private inner class** | Single complex functionality needing state |
| **Separate class** | Complex, reusable handler across classes |

**Lambda expressions are PREFERRED for most simple cases!** They're clean, modern, and easy to read.

## Mouse Events

JavaFX supports many different mouse events beyond just clicking buttons!

**Common Mouse Events**:
- `setOnMousePressed()` - Mouse button pressed down
- `setOnMouseReleased()` - Mouse button released
- `setOnMouseClicked()` - Complete click (press + release)
- `setOnMouseMoved()` - Mouse moved (no button)
- `setOnMouseDragged()` - Mouse moved WITH button pressed
- `setOnMouseEntered()` - Mouse enters component
- `setOnMouseExited()` - Mouse exits component

You can get mouse coordinates using `e.getX()` and `e.getY()`!

````python
/* Mouse event example - DON'T RUN */

pane.setOnMouseClicked(e -> {
    circle.setCenterX(e.getX());
    circle.setCenterY(e.getY());
    System.out.println("Circle moved to: " + e.getX() + ", " + e.getY());
});

pane.setOnMouseEntered(e -> {
    pane.setStyle("-fx-background-color: lightgreen;");
});
````

## Key Events

Keyboard events let us handle key presses for interactive applications!

**Key Event Types**:
- `setOnKeyPressed()` - Key pressed down
- `setOnKeyReleased()` - Key released
- `setOnKeyTyped()` - Character typed

**KeyCode enumeration** - special constants for keys like arrows, Enter, Escape, etc.

````python
/* Key event example - DON'T RUN */

pane.setOnKeyPressed(e -> {
    switch (e.getCode()) {
        case UP:
            circle.setCenterY(circle.getCenterY() - 10);
            break;
        case DOWN:
            circle.setCenterY(circle.getCenterY() + 10);
            break;
        case LEFT:
            circle.setCenterX(circle.getCenterX() - 10);
            break;
        case RIGHT:
            circle.setCenterX(circle.getCenterX() + 10);
            break;
        case ESCAPE:
            System.exit(0);
            break;
    }
});

// Don't forget to request focus!
pane.requestFocus();
````

## Animations

JavaFX provides animation classes for moving objects over time!

**Animation class hierarchy**:
- `Animation` (abstract base class)
  - `PathTransition` - Move object along a path
  - `FadeTransition` - Change opacity over time
  - `Timeline` - Frame-by-frame animation

**Key animation properties**:
- `duration` - How long animation takes (using `Duration` class)
- `cycleCount` - Number of times to repeat (use `Timeline.INDEFINITE` for forever)
- `autoReverse` - Whether to reverse direction on alternate cycles
- `rate` - Speed and direction of animation

````python
/* PathTransition animation example - DON'T RUN */

// Create animation
PathTransition pathTransition = new PathTransition();
pathTransition.setDuration(Duration.millis(4000));  // 4 seconds
pathTransition.setPath(circlePath);                 // Path to follow
pathTransition.setNode(rectangle);                  // Object to animate
pathTransition.setCycleCount(Timeline.INDEFINITE);  // Loop forever
pathTransition.setAutoReverse(true);                // Bounce back and forth
pathTransition.play();                              // Start!

// Can pause/resume with mouse events
circlePath.setOnMousePressed(e -> pathTransition.pause());
circlePath.setOnMouseReleased(e -> pathTransition.play());
````
