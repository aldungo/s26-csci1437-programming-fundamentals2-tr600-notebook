### Objects

An **object** represents a unique/distinct entity in the real world. In programming, objects of the *same type* fall under the definition of a **class**. Objects belong to a class, classes define what the objects characteristics and abilities are. 
An **object** has a state and also has the ability to perform behaviors
- state = properties -> in programming, we know these properties as variables (**fields**)
- behavior = actions -> in programming, these are defined by METHODS 

Example: Imagine a Person Class
- we can have many Person objects
- Example of Properties: Name, Age
- What would be an example of Person behavior? Everybody speaks, everybody works 

**Class** - is a .java file, but more considerably a blueprint or template
* your object is an INSTANCE of the class.
* you instantiate objects of a class, with a **constructor**

### More on Constructors...

```java
public Circle()
{
      radius = 1; // by default, i want my radius of any generic circle created to just be 1
}

public Circle(double newRadius) // parameter argument of double data type
{
      radius = newRadius; // the circle's radius field is initially set to the newRadius value passed in
}
```

**Constructors** - specialized methods to construct/instantiate objects
    - these *initialize* an object's state
    - for example above, default vs. initialized
        - radius is either default, as 1
        - or set upon construction, when user creates object 
    - NAME OF CONSTRUCTOR MATCHES THE NAME OF CLASS
    - no return type, not even void
    - in the main method itself where object creation happens, note that the `new` keyword is used


Constructors with no arguments are called **no-args constructors**, where as ones that DO take arguments are called **parameterized constructors**. 

**Note**: how does the `new` keyword work in object creation?
* When an object is newly created, memory is specifically allocated for that object using **reference variables**. (Familiarize yourself with this term already if you haven't; we use similar context with String reference variables, array reference variables, etc.) 


### Objects Are Accessed via Reference Variables 
When we created `myCircle` in the example, it is a REFERENCE TO a Circle object.

Each of these object references has the ability to access the class's (and its own, technically) data and methods, using the **dot operator** - also known as object member access operator.
Fields and methods of Circle class are **INSTANCE** fields and methods, meaning that they belong to each specific instance of a Circle object. 
Instance fields/methods accessed on the **OBJECTS** of the class. Each object has its own field. 


### Primitive vs. Reference Variables
We know variables as named locations in memory, holding some value for us. 

For PRIMITIVE TYPES, the actual value is held at the location.
For REFERENCE TYPES (think arrays, Classes, and more...), the reference (refer to where its at, not the value itself) is held at the location.

```java
int num1 = 25; // a literal numeric value of 25 is held at num1...
int num2 = num1; // ... so when this copy is done, an actual numeric 25 is assigned value to num2

System.out.print(num1); // literal numeric value 25 is here

Circle cir1 = new Circle(7.5); 
Circle cir2 = cir1; // i wonder what happens here...
```

Because objects are reference variables *to the class*, the behavior would be the same of arrays. You can't just directly copy it!
Be careful with this!
