Constructors - special methods in a Class that allow for objects to be created
- may be parameterized constructor or default 
- parameterized: Constructor takes arguments
- no-args constructor allow the value to default
- Constructor matches name of class

**Objects are invoked "created" using the `new` operator; creates/initializes objects

* When newly created objects are instantiated, memory is allocated for it
### Remember that objects are reference variables! This means that we don't point to the value itself (like an integer, which is a primitive data type) but rather we point to the LOCATION

Remember how the dot operator works! It allows an object to access its data and methods (a.k.a. - object member acccess operator)


With objects being reference variables, we must be careful when copying objects!!

*You could accidentally create a duplicate that gets affected when only one changes* 

Remember that Java has a lot of pre-built classes! We use some of these already! Scanner, Math, `Date`

**Visibility Modifiers** - determines the visibility of a class and its members (also known as access specifiers)
- public: access possible from ANY class
- private: access only from within the SAME class -> more secure!!! prevents data tampering and possibility of bugs 

Great class design makes use of "getters" and "setters"
- getters can access a field (return a field's value)
- setters can assign a new value (mutators) 

 ## How can we PASS OBJECTS TO METHODS?

Objects are reference variables... so when they are passed to methods, they are **passed by sharing (reference)**

## The `this` reference

* `this` keyword references to the calling object
* prevents variable shadowing
* GREAT practice to use in setters and constructors
* `this` can call one constructor from another
